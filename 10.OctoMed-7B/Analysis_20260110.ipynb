```python
df2 = pd.read_pickle('/Users/rodrigocarrillo/Documents/Natural Language Processing Projects/Examen Residentado Peru/02.LLMs Results/10.OctoMed-7B/Chunk_2/combined_exam_dataset_OctoMed-7B_chunk2.pickle')
df2
```

```python
df2['letra_respuesta'] = df2['answer_llm'].str.extract(r'Respuesta final:\s?\(?([ABCDE])\)?')
df2
```

```python
df2['letra_respuesta'].value_counts(dropna = False)
```

```python
df2[df2['letra_respuesta'].isna()][['question', 'answer_llm', 'letra_respuesta']]   # Manually verified that these 09 cases do not contain a valid answer.
```

```python
df2 = df2.dropna(subset=['letra_respuesta'])
df2 = df2.reset_index(drop=True)
df2.shape
```

```python
pd.crosstab(df2['correct_answer'], df2['letra_respuesta'], normalize='columns')
```

########################################################################################################################################################################################################

```python
df3 = pd.concat([df, df2], ignore_index=True)
df3
```

```python
df3['is_correct'] = df3['correct_answer'] == df3['letra_respuesta']

results = df3.groupby(['source_file', 'year']).agg(
    count_of_matches=('is_correct', 'sum'),
    total_rows=('is_correct', 'size')
).reset_index()
results['percent_correct'] = (results['count_of_matches'] / results['total_rows']) * 100
results['percent_correct'] = results['percent_correct'].round(2)
# results['performance'] = (results['percent_correct'] >= 52).astype(int)
# results[results['performance'] == 1]
results
```

########################################################################################################################################################################################################

```python
results['model_name'] = 'OctoMed-7B'
results.to_pickle('/Users/rodrigocarrillo/Documents/Natural Language Processing Projects/Examen Residentado Peru/02.LLMs Results/10.OctoMed-7B/_Results.pickle')
```
