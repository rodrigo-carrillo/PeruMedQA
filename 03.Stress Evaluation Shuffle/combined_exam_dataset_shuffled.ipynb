```python
def shuffle_quiz_options(input_pickle, output_pickle):
    """
    Randomly shuffle multiple-choice options for each question and update the correct answer.
    
    Parameters:
    - input_pickle: path to input pickle file
    - output_pickle: path to output pickle file
    """
    # Read the pickle file
    df = pd.read_pickle(input_pickle)
    
    # Define the option columns
    option_cols = ['option_A', 'option_B', 'option_C', 'option_D', 'option_E']
    option_labels = ['A', 'B', 'C', 'D', 'E']
    
    # Create new columns for shuffled options
    shuffled_data = []
    
    for idx, row in df.iterrows():
        # Get the current correct answer label
        correct_label = row['correct_answer'].strip().upper()
        
        # Create a list of (label, option_text) tuples
        options = [(label, row[col]) for label, col in zip(option_labels, option_cols)]
        
        # Find the correct option text
        correct_idx = option_labels.index(correct_label)
        correct_text = options[correct_idx][1]
        
        # Extract just the option texts
        option_texts = [opt[1] for opt in options]
        
        # Shuffle the option texts
        random.shuffle(option_texts)
        
        # Find the new position of the correct answer
        new_correct_idx = option_texts.index(correct_text)
        new_correct_label = option_labels[new_correct_idx]
        
        # Create a new row with shuffled options
        new_row = row.copy()
        for i, col in enumerate(option_cols):
            new_row[col] = option_texts[i]
        new_row['correct_answer'] = new_correct_label
        
        shuffled_data.append(new_row)
    
    # Create new dataframe with shuffled data
    df_shuffled = pd.DataFrame(shuffled_data)
    
    # Save to pickle
    df_shuffled.to_pickle(output_pickle)
    print(f"Successfully shuffled {len(df_shuffled)} questions.")
    print(f"Output saved to: {output_pickle}")
    
    return df_shuffled
```

```python
input_file = "/Users/rodrigocarrillo/Documents/Natural Language Processing Projects/Examen Residentado Peru/01.Datasets/combined_exam_dataset.pkl"  # Change this to your input file name
output_file = "/Users/rodrigocarrillo/Documents/Natural Language Processing Projects/Examen Residentado Peru/02.LLMs Results/Reorder Options/combined_exam_dataset_shuffled.pkl"  # Change this to your desired output file name

df_result = shuffle_quiz_options(input_file, output_file)
```
