```python
import pickle
import os

# Path to the input dataset
input_path = "/Users/rodrigocarrillo/Documents/Natural Language Processing Projects/Examen Residentado Peru/01.Datasets/combined_exam_dataset.pkl"

# Output directory
output_dir = "/Users/rodrigocarrillo/Documents/Natural Language Processing Projects/Examen Residentado Peru/02.LLMs Results/10.OctoMed-7B"

# Create output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# Load the dataset
print("Loading dataset...")
with open(input_path, 'rb') as f:
    dataset = pickle.load(f)

print(f"Dataset type: {type(dataset)}")
print(f"Dataset length: {len(dataset)}")

# Split the dataset into two equal chunks
mid_point = len(dataset) // 2

chunk_1 = dataset[:mid_point]
chunk_2 = dataset[mid_point:]

print(f"\nChunk 1 size: {len(chunk_1)}")
print(f"Chunk 2 size: {len(chunk_2)}")

# Save the chunks
chunk_1_path = os.path.join(output_dir, "chunk_1.pkl")
chunk_2_path = os.path.join(output_dir, "chunk_2.pkl")

print(f"\nSaving chunk 1 to {chunk_1_path}...")
with open(chunk_1_path, 'wb') as f:
    pickle.dump(chunk_1, f)

print(f"Saving chunk 2 to {chunk_2_path}...")
with open(chunk_2_path, 'wb') as f:
    pickle.dump(chunk_2, f)

print("\nDone! Both chunks saved successfully.")
```
